# Object-level Delphes analysis

This notebook opens a completed Finding Z detector run with `uproot`, Awkward Array, and Vector. Standard runs contain reconstructed physics objects; advanced runs additionally contain particles, tracks, towers, and particle-flow collections.

```python
import awkward as ak

from findingz.delphes import list_runs, open_run

runs = list_runs()
runs
```

Choose a `run_id` from the table above. The final row is used here only as a convenient default.

```python
run_id = runs.iloc[-1]["run_id"]
events = open_run(run_id)
print(events.event_count, "events")
print(events.available_collections)
```

```python
good_muons = events.muons[(events.muons.pt > 20) & (abs(events.muons.eta) < 2.4)]
good_jets = events.jets[(events.jets.pt > 30) & (abs(events.jets.eta) < 2.5)]
selected = (ak.num(good_muons) >= 2) & (ak.num(good_jets) <= 1)
print(ak.sum(selected), "events pass the example selection")
```

For an advanced-profile run, low-level collections are available lazily as `events.particles`, `events.tracks`, `events.towers`, or through `events.collection("eflow_tracks")`. Requesting these from a standard run gives an explicit error rather than silently returning empty objects.
